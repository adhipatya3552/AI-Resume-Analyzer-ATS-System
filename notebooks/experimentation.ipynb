```python
import sys
import os

sys.path.append(
    os.path.abspath("..")
)

import pandas as pd

from src.pdf_parser import extract_text
from src.preprocess import clean_text
from src.skill_extractor import extract_skills
from src.similarity_engine import calculate_similarity
from src.job_loader import (
    load_jobs,
    get_job_titles,
    get_job_description
)
```

```python
resume_path = "data/resumes/software_engineering_manager.pdf"

resume_text = extract_text(resume_path)

print(resume_text[:2000])
```

```python
cleaned_resume = clean_text(resume_text)

print(cleaned_resume[:1000])
```

```python
skills = extract_skills(cleaned_resume)

print("Detected Skills:")
print(skills)
```

jobs_df = load_jobs("/data/jobs/job_descriptions.csv")

jobs_df.head()

```python
print(jobs_df.columns)
```

```python
job_titles = get_job_titles(jobs_df)

print(job_titles[:20])
```

```python
selected_job = job_titles[0]

print(selected_job)
```

```python
job_description = get_job_description(
    jobs_df,
    selected_job
)

print(job_description[:3000])
```

```python
cleaned_job = clean_text(job_description)

print(cleaned_job[:1000])
```

```python
similarity_score = calculate_similarity(cleaned_resume,cleaned_job)

print(f"ATS Match Score: {similarity_score}%")
```

```python
job_skills = extract_skills(cleaned_job)

print(job_skills)
```

```python
from src.recommendation_engine import missing_skills

missing = missing_skills(skills,job_skills)

print("Missing Skills:")
print(missing)
```

```python
comparison_df = pd.DataFrame({

    "Resume Skills": pd.Series(skills),

    "Job Skills": pd.Series(job_skills)

})

comparison_df
```

```python
import matplotlib.pyplot as plt

skill_counts = {}

for skill in skills:
    skill_counts[skill] = 1

plt.figure(figsize=(10, 5))

plt.bar(
    skill_counts.keys(),
    skill_counts.values()
)

plt.xticks(rotation=45)

plt.title("Resume Skills")

plt.show()
```
